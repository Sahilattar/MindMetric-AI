```python
top_countries = df['Country'].value_counts().index[:10].tolist()
def GrpCountry(country):
  if country in top_countries:
      return country
  else:
      return 'Other'
```

```python
df['GrpCountry'] = df['Country'].apply(GrpCountry)
```

```python
from sklearn.model_selection import train_test_split
skwed_col = ["Study_Hours"]
other_col = ['Age','Avg_Daily_Usage_Hours', 'Daily_Unlocks','Physical_Activity_Hours', 'Sleep_Hours_Per_Night',]
ordinal_col = ["Stress_Level"]
normal_col = ['Gender','Academic_Level', 'Most_Used_Platform','Purpose_Of_Use','GrpCountry']
feature_col = skwed_col+other_col+ordinal_col+normal_col
X = df[feature_col]
y = df['Mental_Health_Score']

X_train, X_test,y_train, y_test= train_test_split(X,y, test_size=0.30, random_state=42)
```

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import FunctionTransformer,StandardScaler,OrdinalEncoder,OneHotEncoder

skew_pipeline = Pipeline(steps=[
    ('log_transfer',FunctionTransformer(np.log1p)),
    ('scaler',StandardScaler())
])

other_pipeline = Pipeline(steps=[
    ('scaler',StandardScaler())
])

ordinal_pipeline = Pipeline(steps=[
    ('encode',OrdinalEncoder(categories=[['Low','Medium','High','Very High']]))
])

normal_pipeline = Pipeline(steps=[
    ('encode',OneHotEncoder(handle_unknown='ignore'))
])
```

```python
from sklearn.compose import ColumnTransformer

preprocesser=ColumnTransformer(transformers=[
    ("skewd_pipeline", skew_pipeline, skwed_col),
    ("other_pipeline", other_pipeline, other_col),
    ("ordinal_pipeline", ordinal_pipeline, ordinal_col),
    ("normal_pipeline", normal_pipeline, normal_col)
])
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
lr_pipeline = Pipeline(steps=[
     ('preprocesser', preprocesser),
     ('regression', LinearRegression())
 ])

lr_pipeline.fit(X_train, y_train)
lr_preds_test = lr_pipeline.predict(X_test)
lr_preds_train = lr_pipeline.predict(X_train)

lr_r2_test = r2_score(y_test, lr_preds_test)
lr_r2_train = r2_score(y_train, lr_preds_train)
lr_r2_mae = mean_absolute_error(y_test, lr_preds)

print(lr_r2_test)
print(lr_r2_train)
print(lr_r2_mae)
```

```python
from sklearn.model_selection import RandomizedSearchCV
param_grid = {
    'random forest__n_estimators' : [100,200,300],
    'random forest__max_depth' : [5,10,15],
    'random forest__min_samples_split' : [2,5,10],
    'random forest__min_samples_leaf' : [1,2,4]
}

random_search = RandomizedSearchCV(
    estimator=rf_pipeline,
    param_distributions=param_grid,
    n_iter=15,
    cv=5,
    scoring='r2',
    random_state=42,
    n_jobs=-1
)

random_search.fit(X_train,y_train)
```
